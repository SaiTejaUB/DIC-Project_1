```python
raw_df = spark.read.csv(
    "/Volumes/workspace/default/uploads/US_Accidents_March23.csv",
    header=True,
    inferSchema=True
)

# Sanitize column names by replacing invalid characters
for col in raw_df.columns:
    raw_df = raw_df.withColumnRenamed(col, col.replace("(", "_").replace(")", "").replace(" ", "_"))

raw_df.write.format("delta").mode("overwrite").saveAsTable("bronze_accidents")
print(f"Bronze done! Rows: {raw_df.count()}")
raw_df.show(5)
```
