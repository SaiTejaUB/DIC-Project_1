```python
# Bronze Layer - Combined Accidents + HPMS Dataset
# Load raw CSV as-is into Delta Table

raw_path = "/Volumes/workspace/default/uploads/us_accidents_hpms_final_100m.csv"
bronze_table = "workspace.default.bronze_accidents_hpms"

raw_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(raw_path)

print(f"Loaded {raw_df.count()} rows")
raw_df.printSchema()
```

```python
# Rename problem columns before saving to Delta
raw_df = raw_df \
    .withColumnRenamed("Distance(mi)", "Distance_mi") \
    .withColumnRenamed("Temperature(F)", "Temperature_F") \
    .withColumnRenamed("Humidity(%)", "Humidity_pct") \
    .withColumnRenamed("Pressure(in)", "Pressure_in") \
    .withColumnRenamed("Visibility(mi)", "Visibility_mi") \
    .withColumnRenamed("Wind_Speed(mph)", "Wind_Speed_mph") \
    .withColumnRenamed("Precipitation(in)", "Precipitation_in")

# save as Bronze Delta Table
raw_df.write.format("delta").mode("overwrite").saveAsTable("workspace.default.bronze_accidents_hpms")
print(f"Bronze layer done! Rows: {raw_df.count()}")
raw_df.show(5)
```
